```python
import numpy as np

print("ПРЕДОБРАБОТКА ДАННЫХ")

# Убираем служебные столбцы (город, штат, коды)
columns_to_drop = ['communityname', 'State', 'countyCode', 'communityCode', 'fold']
df = df.drop(columns=columns_to_drop)
print(f"После удаления служебных столбцов: {df.shape}")

target_column = 'ViolentCrimesPerPop'

# Убираем другие "целевые" столбцы, чтобы модель их не подсматривала
other_targets = [
    'murders', 'murdPerPop', 'rapes', 'rapesPerPop', 'robberies',
    'robbbPerPop', 'assaults', 'assaultPerPop', 'burglaries', 'burglPerPop',
    'larcenies', 'larcPerPop', 'autoTheft', 'autoTheftPerPop', 'arsons',
    'arsonsPerPop', 'nonViolPerPop'
]

X = df.drop(columns=[target_column] + other_targets)
y = df[target_column]

print(f"Размер X: {X.shape}")
print(f"Размер y: {y.shape}")

# Проверка на текстовые столбцы которых не должно быть
non_numeric = X.select_dtypes(exclude=[np.number]).columns.tolist()
if len(non_numeric) == 0:
    print("\nВсе признаки числовые!")
else:
    print(f"\nНечисловые столбцы: {non_numeric}")

# Удаляем строки с пропусками сразу в X и y
data_clean = pd.concat([X, y], axis=1).dropna()
X_clean = data_clean.drop(columns=[target_column])
y_clean = data_clean[target_column]

print(f"\nПосле удаления пропусков")
print(f"X_clean: {X_clean.shape}")
print(f"y_clean: {y_clean.shape}")
print(f"Потеряно строк: {X.shape[0] - X_clean.shape[0]}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression

print("\nКОРРЕЛЯЦИЯ И МУЛЬТИКОЛЛИНЕАРНОСТЬ")

# Смотрим, какие признаки сильнее всего связаны с целевой переменной
correlations = X_clean.corrwith(y_clean).abs().sort_values(ascending=False)

print("\nТоп признаков по корреляции с целевой:")
print(correlations.head(15))

selected_features = [
    'PctKids2Par',
    'racepctblack',
    'TotalPctDiv',
    'PctPopUnderPov',
    'PctUnemployed',
    'PctHousOwnOcc',
    'medIncome'
]

X_selected = X_clean[selected_features]
y_selected = y_clean

print(f"\nВыбрано {len(selected_features)} признаков:")
for f in selected_features:
    print(f"   {f}: {X_selected[f].corr(y_selected):+.4f}")

corr_matrix = pd.concat([X_selected, y_selected], axis=1).corr()

plt.figure(figsize=(11, 9))
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap='RdBu_r',
    center=0,
    vmin=-1, vmax=1,
    square=True,
    linewidths=0.5,
    linecolor='white',
    cbar_kws={'shrink': 0.8, 'label': 'Коэффициент корреляции'},
    annot_kws={'size': 9}
)
plt.title('Матрица корреляций', fontsize=14, pad=15)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

def calculate_vif(X):
    vif_data = pd.DataFrame()
    vif_data["Признак"] = X.columns
    vif_values = []
    for col in X.columns:
        y_i = X[col]
        X_i = X.drop(columns=[col])
        model = LinearRegression().fit(X_i, y_i)
        r_sq = model.score(X_i, y_i)
        vif = 1 / (1 - r_sq) if r_sq < 1 else np.inf
        vif_values.append(vif)
    vif_data["VIF"] = vif_values
    return vif_data


vif_result = calculate_vif(X_selected)
print("\nVIF (если < 10 — мультиколлинеарности нет):")
print(vif_result.to_string(index=False))

max_vif = vif_result['VIF'].max()
if max_vif < 10:
    print(f"\nМультиколлинеарности нет (max VIF = {max_vif:.2f})")
else:
    print(f"\nЕсть мультиколлинеарность (max VIF = {max_vif:.2f})")
```

```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_percentage_error

print("ЛИНЕЙНАЯ И ГРЕБНЕВАЯ РЕГРЕССИЯ")

print(f"Target непрерывная используем линейную регрессию\n")

# Проверяем масштаб признаков
print("Масштаб признаков (min / max)")
for f in selected_features:
    print(f"  {f}: {X_selected[f].min():.2f} / {X_selected[f].max():.2f}")
print("\n Признаки в разных масштабах нужна стандартизация!\n")

X_train, X_test, y_train, y_test = train_test_split(
    X_selected, y_selected, test_size=0.3, random_state=42
)
print(f"Train: {X_train.shape}")
print(f"Test:  {X_test.shape}")

# Стандартизация
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"\nСреднее после стандартизации (должно быть около 0):")
print(f"  {np.round(X_train_scaled.mean(axis=0), 4)}")
print(f"Стд. отклонение (должно быть около 1):")
print(f"  {np.round(X_train_scaled.std(axis=0), 4)}\n")

# Функция для расчёта метрик
def evaluate_model(name, y_true, y_pred):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    mask = y_true != 0
    mape = mean_absolute_percentage_error(y_true[mask], y_pred[mask]) * 100
    print(f"\n--- {name} ---")
    print(f"  RMSE: {rmse:.4f}")
    print(f"  R^2:  {r2:.4f}")
    print(f"  MAPE: {mape:.2f}%")
    return {'Модель': name, 'RMSE': round(rmse, 4), 'R^2': round(r2, 4), 'MAPE(%)': round(mape, 2)}

results = []

# Линейная регрессия на стандартизированных данных
lin_reg = LinearRegression()
lin_reg.fit(X_train_scaled, y_train)
y_pred_lin = lin_reg.predict(X_test_scaled)
results.append(evaluate_model("Linear Regression (исходные)", y_test, y_pred_lin))

# Кросс-валидация
cv_scores = cross_val_score(lin_reg, X_train_scaled, y_train, cv=5, scoring='r2')
print(f"\nКросс-валидация 5-fold R^2: {cv_scores.mean():.4f} ± {cv_scores.std():.4f}")

# Ridge-регрессия на стандартизированных данных
ridge_reg = Ridge(alpha=1.0, random_state=42)
ridge_reg.fit(X_train_scaled, y_train)
y_pred_ridge = ridge_reg.predict(X_test_scaled)
results.append(evaluate_model("Ridge Regression (исходные)", y_test, y_pred_ridge))
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

print("PCA")

# Стандартизируем признаки перед PCA
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_selected)

# Оставляем компоненты, которые объясняют 95% дисперсии
pca = PCA(n_components=0.95, random_state=42)
X_pca = pca.fit_transform(X_scaled)

print(f"Количество компонент: {pca.n_components_}")
print(f"Объяснённая дисперсия: {pca.explained_variance_ratio_.sum():.4f}")

print("\nДисперсия по компонентам")
for i, var in enumerate(pca.explained_variance_ratio_):
    print(f"  PC{i+1}: {var*100:.2f}%")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].bar(range(1, len(pca.explained_variance_ratio_)+1),
            pca.explained_variance_ratio_, color='steelblue', edgecolor='black')
axes[0].plot(range(1, len(pca.explained_variance_ratio_)+1),
             np.cumsum(pca.explained_variance_ratio_), 'ro-')
axes[0].set_xlabel('Компонента')
axes[0].set_ylabel('Доля дисперсии')
axes[0].set_title('Scree Plot')
axes[0].grid(True, alpha=0.3)

loadings = pd.DataFrame(
    pca.components_.T,
    columns=[f'PC{i+1}' for i in range(pca.n_components_)],
    index=selected_features
)
sns.heatmap(loadings, annot=True, fmt=".2f", cmap='coolwarm', center=0, ax=axes[1])
axes[1].set_title('Значимость признаков в компонентах')

plt.tight_layout()
plt.show()
```

```python
print("РЕГРЕССИЯ НА PCA")

X_train_pca, X_test_pca, y_train_pca, y_test_pca = train_test_split(
    X_pca, y_selected, test_size=0.3, random_state=42
)

# Линейная регрессия на компонентах
lin_reg_pca = LinearRegression()
lin_reg_pca.fit(X_train_pca, y_train_pca)
y_pred_pca = lin_reg_pca.predict(X_test_pca)
results.append(evaluate_model("Linear Regression (PCA)", y_test_pca, y_pred_pca))

# Кросс-валидация
cv_pca = cross_val_score(lin_reg_pca, X_pca, y_selected, cv=5, scoring='r2')
print(f"\nКросс-валидация 5-fold R^2: {cv_pca.mean():.4f} ± {cv_pca.std():.4f}")

# Ridge на компонентах
ridge_pca = Ridge(alpha=1.0, random_state=42)
ridge_pca.fit(X_train_pca, y_train_pca)
y_pred_ridge_pca = ridge_pca.predict(X_test_pca)
results.append(evaluate_model("Ridge Regression (PCA)", y_test_pca, y_pred_ridge_pca))

print("\nФинальное сравнение моделей")
results_df = pd.DataFrame(results).sort_values('R^2', ascending=False)
print(results_df.to_string(index=False))

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

axes[0].scatter(y_test, y_pred_lin, alpha=0.4, color='steelblue')
axes[0].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
axes[0].set_xlabel('Реальные значения')
axes[0].set_ylabel('Предсказания')
axes[0].set_title(f'Linear (исходные) R^2 = {r2_score(y_test, y_pred_lin):.3f}')

axes[1].scatter(y_test_pca, y_pred_pca, alpha=0.4, color='coral')
axes[1].plot([y_test_pca.min(), y_test_pca.max()],
             [y_test_pca.min(), y_test_pca.max()], 'r--', lw=2)
axes[1].set_xlabel('Реальные значения')
axes[1].set_ylabel('Предсказания')
axes[1].set_title(f'Linear (PCA) R^2 = {r2_score(y_test_pca, y_pred_pca):.3f}')

plt.tight_layout()
plt.show()

metrics_df = results_df.set_index('Модель')[['RMSE', 'R^2', 'MAPE(%)']]
metrics_df.plot(kind='bar', subplots=True, figsize=(12, 8), legend=False,
                color=['steelblue', 'seagreen', 'coral'])
plt.suptitle('Сравнение метрик моделей', fontsize=14)
plt.tight_layout()
plt.show()

print("\n ИТОГОВЫЙ АНАЛИЗ РЕЗУЛЬТАТОВ")

best_model = results_df.iloc[0]
worst_model = results_df.iloc[-1]

print(f"\nЛучшая модель: {best_model['Модель']}")
print(f"   R^2 = {best_model['R^2']}, RMSE = {best_model['RMSE']}, MAPE = {best_model['MAPE(%)']}%")

print(f"\nХудшая модель: {worst_model['Модель']}")
print(f"   R^2 = {worst_model['R^2']}, RMSE = {worst_model['RMSE']}, MAPE = {worst_model['MAPE(%)']}%")

print(f"\nРазница R^2 между лучшей и худшей: {best_model['R^2'] - worst_model['R^2']:.4f}")

print("\nСравнение: исходные признаки vs PCA")
lin_orig = results_df[results_df['Модель'] == 'Linear Regression (исходные)'].iloc[0]
lin_pca = results_df[results_df['Модель'] == 'Linear Regression (PCA)'].iloc[0]

print(f"  Linear (исходные): R^2 = {lin_orig['R^2']}, RMSE = {lin_orig['RMSE']}, MAPE = {lin_orig['MAPE(%)']}%")
print(f"  Linear (PCA):      R^2 = {lin_pca['R^2']}, RMSE = {lin_pca['RMSE']}, MAPE = {lin_pca['MAPE(%)']}%")
print(f"  Изменение R^2: {lin_pca['R^2'] - lin_orig['R^2']:+.4f}")
```
