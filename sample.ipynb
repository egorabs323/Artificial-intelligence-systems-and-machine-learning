```python
import numpy as np

print("ПРЕДОБРАБОТКА ДАННЫХ")

# Убираем служебные столбцы (город, штат, коды)
columns_to_drop = ['communityname', 'State', 'countyCode', 'communityCode', 'fold']
df = df.drop(columns=columns_to_drop)
print(f"После удаления служебных столбцов: {df.shape}")

target_column = 'ViolentCrimesPerPop'

# Убираем другие "целевые" столбцы, чтобы модель их не подсматривала
other_targets = [
    'murders', 'murdPerPop', 'rapes', 'rapesPerPop', 'robberies',
    'robbbPerPop', 'assaults', 'assaultPerPop', 'burglaries', 'burglPerPop',
    'larcenies', 'larcPerPop', 'autoTheft', 'autoTheftPerPop', 'arsons',
    'arsonsPerPop', 'nonViolPerPop'
]

X = df.drop(columns=[target_column] + other_targets)
y = df[target_column]

print(f"Размер X: {X.shape}")
print(f"Размер y: {y.shape}")

# Проверка на текстовые столбцы которых не должно быть
non_numeric = X.select_dtypes(exclude=[np.number]).columns.tolist()
if len(non_numeric) == 0:
    print("\nВсе признаки числовые!")
else:
    print(f"\nНечисловые столбцы: {non_numeric}")

# Удаляем строки с пропусками сразу в X и y
data_clean = pd.concat([X, y], axis=1).dropna()
X_clean = data_clean.drop(columns=[target_column])
y_clean = data_clean[target_column]

print(f"\nПосле удаления пропусков")
print(f"X_clean: {X_clean.shape}")
print(f"y_clean: {y_clean.shape}")
print(f"Потеряно строк: {X.shape[0] - X_clean.shape[0]}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression

print("\nКОРРЕЛЯЦИЯ И МУЛЬТИКОЛЛИНЕАРНОСТЬ")

# Смотрим, какие признаки сильнее всего связаны с целевой переменной
correlations = X_clean.corrwith(y_clean).abs().sort_values(ascending=False)

print("\nТоп признаков по корреляции с целевой:")
print(correlations.head(15))

selected_features = [
    'PctKids2Par',
    'racepctblack',
    'TotalPctDiv',
    'PctPopUnderPov',
    'PctUnemployed',
    'PctHousOwnOcc',
    'medIncome'
]

X_selected = X_clean[selected_features]
y_selected = y_clean

print(f"\nВыбрано {len(selected_features)} признаков:")
for f in selected_features:
    print(f"   {f}: {X_selected[f].corr(y_selected):+.4f}")

corr_matrix = pd.concat([X_selected, y_selected], axis=1).corr()

plt.figure(figsize=(11, 9))
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap='RdBu_r',
    center=0,
    vmin=-1, vmax=1,
    square=True,
    linewidths=0.5,
    linecolor='white',
    cbar_kws={'shrink': 0.8, 'label': 'Коэффициент корреляции'},
    annot_kws={'size': 9}
)
plt.title('Матрица корреляций', fontsize=14, pad=15)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

def calculate_vif(X):
    vif_data = pd.DataFrame()
    vif_data["Признак"] = X.columns
    vif_values = []
    for col in X.columns:
        y_i = X[col]
        X_i = X.drop(columns=[col])
        model = LinearRegression().fit(X_i, y_i)
        r_sq = model.score(X_i, y_i)
        vif = 1 / (1 - r_sq) if r_sq < 1 else np.inf
        vif_values.append(vif)
    vif_data["VIF"] = vif_values
    return vif_data


vif_result = calculate_vif(X_selected)
print("\nVIF (если < 10 — мультиколлинеарности нет):")
print(vif_result.to_string(index=False))

max_vif = vif_result['VIF'].max()
if max_vif < 10:
    print(f"\nМультиколлинеарности нет (max VIF = {max_vif:.2f})")
else:
    print(f"\nЕсть мультиколлинеарность (max VIF = {max_vif:.2f})")
```
